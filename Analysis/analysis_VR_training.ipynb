# VR Training Analysis

Processes and plots the VR-based perceptual training sessions: for each training day,
a noise-contrast staircase estimates how much dichoptic noise disrupts the amblyopic eye's
perception (see `VR_processing.py`), averaged over the last 6 reversals per run. Tracks
this threshold's progress across the ~15 training days for patients and controls, plus
reaction times, catch-trial performance, and per-participant training curves.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def plot_noise_and_delta_by_day(
    df,
    participant_col=None,
    participant_value=None,
    date_col='date_to_number',
    run_col='run_number',
    noise_col='noise_contrast',
    delta_col='delta',
    days_to_plot=None,          # list of date identifiers (in desired order) or None for first_n_days
    first_n_days=5,             # used if days_to_plot is None
    runs_per_day=20,
    day_labels_prefix='Day',
    gap_between_days=2,         # gap units to visually separate days on x axis
    smoothing_window=False,      # int: moving average window in runs (True -> default 5; None/False = no smoothing)
    show_markers=True,
    figsize=(12,5),
    title='Noise Contrast and Delta across Days',
    savepath=None,              # if provided, save figure to this file (png/pdf/etc.)
    noise_ylim=(0, 100),        # tuple or None; set to (0,100) to force noise axis
    delta_ylim=(0, 30)          # tuple or None; set to (0,30) to force delta axis
):
    """
    Plot noise_contrast and delta vs run across multiple days on a single figure.
    - Major x-ticks: one per day (centered, labeled "Day 1", ... or custom via days_to_plot)
    - Minor x-ticks: one per run (0..runs_per_day-1) for each day

    New params:
      noise_ylim: (min,max) or None to disable forcing the noise y-axis limits
      delta_ylim: (min,max) or None to disable forcing the delta y-axis limits

    Note on smoothing_window:
      - If smoothing_window is True -> uses window=5
      - If smoothing_window is an int > 1 -> uses that window
      - If smoothing_window is None or False -> no smoothing
    """
    # ---- Input checks and filtering ----
    df = df.copy()
    if participant_col is not None and participant_value is not None:
        df = df[df[participant_col] == participant_value].copy()
        if df.empty:
            raise ValueError("No rows found after filtering by participant.")

    # ensure numeric ints for mapping
    df[date_col] = df[date_col].astype(int)
    df[run_col] = df[run_col].astype(int)

    unique_days = sorted(df[date_col].unique())
    if days_to_plot is None:
        days_to_plot = unique_days[:first_n_days]
    else:
        # keep only days that exist in the data, preserve order given by user
        days_to_plot = [d for d in days_to_plot if d in unique_days]

    if len(days_to_plot) == 0:
        raise ValueError("No matching days to plot. Check your 'days_to_plot' or data.")

    # Map days to indices 0..n_days-1
    day_to_idx = {d: i for i, d in enumerate(days_to_plot)}
    df = df[df[date_col].isin(days_to_plot)].copy()
    df['day_idx'] = df[date_col].map(day_to_idx)
    span_per_day = runs_per_day + gap_between_days
    df['x_pos'] = df['day_idx'] * span_per_day + df[run_col]

    # Interpret smoothing_window parameter
    if smoothing_window is True:
        smoothing_window_val = 5
    elif isinstance(smoothing_window, int) and smoothing_window > 1:
        smoothing_window_val = smoothing_window
    else:
        smoothing_window_val = None

    # Optionally smooth columns (moving average over runs — applied per day)
    if smoothing_window_val is not None and smoothing_window_val > 1:
        sm_noise = []
        sm_delta = []
        for d in days_to_plot:
            sub = df[df[date_col] == d].sort_values(run_col)
            # pad/align to runs 0..runs_per_day-1 if some runs missing: use reindex
            idx = range(runs_per_day)
            sub = sub.set_index(run_col).reindex(idx)
            smn = sub[noise_col].rolling(window=smoothing_window_val, min_periods=1, center=True).mean()
            smd = sub[delta_col].rolling(window=smoothing_window_val, min_periods=1, center=True).mean()
            # put back into lists with corresponding x positions
            base = day_to_idx[d] * span_per_day
            for run in idx:
                x = base + run
                sm_noise.append({'x_pos': x, noise_col: smn.loc[run], date_col: d, run_col: run})
                sm_delta.append({'x_pos': x, delta_col: smd.loc[run], date_col: d, run_col: run})
        # build smoothed DataFrame for plotting
        sm_noise_df = pd.DataFrame(sm_noise).dropna(subset=[noise_col])
        sm_delta_df = pd.DataFrame(sm_delta).dropna(subset=[delta_col])
    else:
        sm_noise_df = None
        sm_delta_df = None

    # ---- Plotting ----
    fig, ax1 = plt.subplots(figsize=figsize)
    ax2 = ax1.twinx()

    # Plot raw points/lines (kept your explicit colors)
    if show_markers:
        l1 = ax1.plot(df['x_pos'], df[noise_col], marker='o', color='blue', linestyle='-', label=noise_col)
        l2 = ax2.plot(df['x_pos'], df[delta_col], marker='s', color='red', linestyle='--', label=delta_col)
    else:
        l1 = ax1.plot(df['x_pos'], df[noise_col], linestyle='-', color='blue', label=noise_col)
        l2 = ax2.plot(df['x_pos'], df[delta_col], linestyle='--', color='red', label=delta_col)

    # Plot smoothed lines if requested
    if sm_noise_df is not None:
        ax1.plot(sm_noise_df['x_pos'], sm_noise_df[noise_col], color='blue', linestyle='-', linewidth=2, label=f'{noise_col}_smoothed')
    if sm_delta_df is not None:
        ax2.plot(sm_delta_df['x_pos'], sm_delta_df[delta_col], color='red', linestyle='--', linewidth=2, label=f'{delta_col}_smoothed')

    # Major ticks: centered per day block
    n_days = len(days_to_plot)
    major_positions = [i * span_per_day + (runs_per_day - 1) / 2.0 for i in range(n_days)]
    # Label days either as "Day 1..n" or use the actual date identifiers
    # Label days using the actual values in days_to_plot
    major_labels = [f"{day_labels_prefix} {d}" for d in days_to_plot]

    ax1.set_xticks(major_positions)
    ax1.set_xticklabels(major_labels)

    # Minor ticks for every run across the plotted range
    minor_positions = []
    for i in range(n_days):
        base = i * span_per_day
        minor_positions.extend([base + r for r in range(runs_per_day)])
    ax1.set_xticks(minor_positions, minor=True)
    ax1.tick_params(axis='x', which='minor', length=4)
    ax1.tick_params(axis='x', which='major', length=8)

    # Axis labels and title
    ax1.set_xlabel('Day and Run numbers')
    ax1.set_ylabel('Noise Contrast (%) in blue') #noise_col
    ax2.set_ylabel('Delta between Gabors in red') #delta_col
    plt.title(title)

    # Force y-limits if requested
    if noise_ylim is not None:
        try:
            ax1.set_ylim(noise_ylim)
        except Exception:
            raise ValueError("noise_ylim must be a (min,max) tuple or None.")
    if delta_ylim is not None:
        try:
            ax2.set_ylim(delta_ylim)
        except Exception:
            raise ValueError("delta_ylim must be a (min,max) tuple or None.")

    # Combined legend (kept your upper-right placement)
    lines_1, labels_1 = ax1.get_legend_handles_labels()
    lines_2, labels_2 = ax2.get_legend_handles_labels()
    ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc='upper right')

    # Grid for readability
    ax1.grid(axis='x', which='minor', linestyle=':', linewidth=0.5)
    ax1.grid(axis='y', which='major', linestyle='-', linewidth=0.5, alpha=0.3)

    fig.tight_layout()

    if savepath:
        fig.savefig(savepath, dpi=300, bbox_inches='tight')
        print(f"Saved figure to: {savepath}")

    plt.show()


# ---------------------------
# Example usage (replace demo_df with your real df):
# ---------------------------
# 1) Simple: first 5 days (by order encountered), 20 runs/day, forced y-limits
# plot_noise_and_delta_by_day(df)

# 2) Specific days (e.g., date_to_number values [10,11,12,13,14] in that order)
# plot_noise_and_delta_by_day(df, days_to_plot=[10,11,12,13,14])

# 3) Filter for participant 8771
# plot_noise_and_delta_by_day(df, participant_col='id_participant', participant_value=8771)

# 4) With smoothing (default 5-run window) and save
# plot_noise_and_delta_by_day(df, smoothing_window=True, savepath='noise_delta_days1-5.png')
```

```python
def plot_noise_contrast_delta(df):
    unique_dates = df['run_number'].unique()
    
    for date in sorted(unique_dates):
        subset = df[df['run_number'] == date]
        date = date +1
        fig, ax1 = plt.subplots(figsize=(10, 5))

        # Line plot for noise_contrast
        sns.lineplot(x="trial_number", y="noise_contrast", data=subset, ax=ax1, color="skyblue", alpha=0.7, label='Noise Contrast')

        # Non-linear regression added
        regplot = sns.regplot(
            x="trial_number", 
            y="noise_contrast", 
            data=subset, 
            scatter=False,
            lowess=True,  # this activates locally weighted scatterplot smoothing
            ax=ax1,
            color="navy",
            line_kws={'linestyle':'--'}
        )

        # Add label manually for regplot
        regplot_line = regplot.get_lines()[-1]  # Get the last line plotted
        regplot_line.set_label('Learning Curve')
        
        ax1.set_ylabel("Noise Contrast", color="blue")
        ax1.set_xlabel("Trial Number")
        ax1.set_ylim(0, 110)

        # Plot horizontal line for delta
        if not subset['delta'].empty:
            ax1.axhline(
                y=50-subset['delta'].iloc[0],
                color="red",
                linestyle='--',
                label=f'Distractor Contrast: {50-subset["delta"].iloc[0]}'
            )
            ax1.legend()

        plt.title(f"Noise Contrast and Distractor contrast for Day 4 and Run {date}")
        plt.show()
```
